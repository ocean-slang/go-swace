```python
# Functions to search and open Lansat scenes for SST
'''
From NLSST pipeline - update from there

Functions to search, open, and analyze Landsat scenes for SST.
Search_stac finds the Landsat scene based on user parameters, 
plot_search plots the locations of the landsat scenes from the search,
landsat_to_xarray takes one of those scenes and puts all bands into an xarray,
and create_masks produces cloud/ice/water masks for the scene. Subset_img 
subsets a landsat scene with coordinates that have been reprojected from lat/lon
and may be flipped in which is larger in the pair. Lsat_reproj can be used to reproject
while ensuring x and y pairs don't get flipped (common converting between espg 3031 and wgs84.
'''

def ls_thermal_to_xarray(sceneid, bandNames=None):
    """
    ***changed sceneid.id to sceneid***
    Function to ingest landsat scenes from AWS s3 bucket with alternate href.
    Args:
        catalog: an intake STAC item collection
        bandNames: a list of the bands you want e.g. ['red', 'swir16']
    Returns:
        Xarray of scenes with dimensions time, band, y, x
    """

    # Import to xarray with cloud mask
    # nans are in locations where concat of multiple scenes has expanded the grid
    item = catalog[sceneid.id]

    bands = []
    band_names = []

    if bandNames is None:
        # Get band names
        for k in item.keys():
            M = getattr(item, k).metadata
            if 'eo:bands' in M:
                resol = M['eo:bands'][0]['gsd']
                if resol >= 30: # thermal bands are up sampled from 100 to 30
                    band_names.append(k)
    else:
        band_names = bandNames

    # Add qa band
    if 'qa_pixel' not in band_names:
        band_names.append('qa_pixel')

    # Construct xarray for scene
    for band_name in band_names:
        asset = sceneid.assets[band_name]
        href = asset.extra_fields['alternate']['s3']['href']
        band = xr.open_dataset(href, engine='rasterio', chunks=dict(band=1, x=512, y=512))
        band['band'] = [band_name]
        bands.append(band)
    ls_scene = xr.concat(bands, dim='band')
    ls_scene.coords['id'] = sceneid.id
    ls_scene.coords['time'] = item.metadata['datetime'].strftime('%Y-%m-%dT%H:%M:%S')
    ls_scene = ls_scene['band_data']

    return ls_scene

##########################

def create_masks(ls_scene, cloud_mask=True, ice_mask=False, ocean_mask=False):
    """
    Create cloud, ice, and ocean masks. cloud is 1, ice is 2, ocean is 3, everything else is NaN.
    Args:
        ls_scene: xarray of one LS scenes output from landsat_to_xarray
        cloud_mask: bool to generate cloud mask, default True
        ice_mask: bool to generate ice mask, default False
        ocean_mask: bool to generate ocean mask, default False
    Returns:
        new xarray of LS scene with mask
    """
    
    cloud = []
    ocean = []
    ice = []

    qa = ls_scene.sel(band='qa_pixel').astype('uint16')

    n,c = np.unique(qa, return_counts=True)

    for j in range(len(n)):
        longform = f'{n[j]:016b}'
        if (longform[-7]=='0')|(longform[-3]=='1'): #bit 2 and 6 are for cirrus and clear sky
            cloud.append(n[j])
        if longform[-8:]=='11000000': #bit 6 and 7 give clear sky and water, lower bits need to be 0 
            ocean.append(n[j])
        if longform[-7:]=='1100000': #bit 5 and 6 give ice and clear sky 
            ice.append(n[j])

    if 0 in cloud:
        cloud.remove(0)
    if 1 in cloud:
        cloud.remove(1)

    # mask cloud, ice, and ocean
    if cloud_mask==True:
        # cloud is 2
        mask_c = xr.where(qa.isin(cloud), 1, np.nan)

    if ice_mask==True:
        mask_c = xr.where(qa.isin(ice), 2, mask_c)

    if ocean_mask==True:
        mask_c = xr.where(qa.isin(ocean), 3, mask_c)

    ls_scene.coords['mask'] = (('y', 'x'), mask_c.data)
        
    return ls_scene

##########################

def normalize(array):
    '''
    normalize a dask array so all value are between 0 and 1
    '''
    array_min = array.min(skipna=True)
    array_max = array.max(skipna=True)
    return (array - array_min) / (array_max - array_min)

##########################

def search_stac(url,bbox,timeRange,collection,gjson_outfile):
    # Search STAC API for Landsat images based on a bounding box, date and other metadata if desired
    api = pystac_client.Client.open(url)

    items = api.search(
                bbox = bbox,
                datetime = timeRange,
                # limit = 400,
                collections=collection
            ).item_collection()

    print(f'{len(items)} Landsat 8/9 items')

    # Write a json file that records our search output
    items.save_object(gjson_outfile)
    
    return items

##########################

def plot_search(gf,satellite,colnm):
    # Plot search AOI and frames on a map using Holoviz Libraries (more on these later)
    cols = gf.loc[:,('id',colnm[0],colnm[1],'geometry')]
    alpha = 1/gf.shape[0]**0.5 # transparency scales w number of images

    footprints = cols.hvplot(geo=True, line_color='k', hover_cols=[colnm[0],colnm[1]], alpha=alpha, title=satellite,tiles='ESRI')
    tiles = gv.tile_sources.CartoEco.options(width=700, height=500) 
    labels = gv.tile_sources.StamenLabels.options(level='annotation')
    tiles * footprints * labels
    
    return footprints

##########################

def subset_img(da,polarx,polary):
    '''
    ***Only works for square grid cropping along the orientation of the grid (not when cropping along lat/lon in a 3031 grid
    
    Subset image in xarray to desired coordinates. Because Landsat polar stereo projection can be oriented
    in many different directions, when coordinates to subset an image are reprojected from lat/lon they may get 
    flipped for which is larger in the pair. This function checks to make sure we are getting a proper subset and 
    avoids 0 pixels on the x or y axis. 
    
    Note: Input shape dimensions and dataarray v. dataset changes things so input needs to be a dataarray w 
          2 dimensions (x,y)
    
    Input:
    da = xarray DataArray to be subset
    polarx = x coordinates to subset by in polar stereographic projection
    polary = y coordinates to subset by in polar stereographic projection
    
    Output:
    ls_sub = subset xarray DataArray
    
    '''
    # ***Landsat shape dimensions are one fewer than they are for LandsatCalibration [0,1] not [1,2], no .to_array() or Band
    ls_sub = da.sel(y=slice(polary[1],polary[0]),x=slice(polarx[0],polarx[1]))

    # Check for right dimensions because y order changes sometimes
    if (ls_sub.x.shape[0]==0) & (ls_sub.y.shape[0]==0):
        print ('L8 x and y shapes are 0')
        ls_sub = da.sel(y=slice(polary[0],polary[1]),x=slice(polarx[1],polarx[0]))
    elif ls_sub.y.shape[0]==0:
        print ('L8 y shape is 0')
        ls_sub = da.sel(y=slice(polary[0],polary[1]),x=slice(polarx[0],polarx[1]))
    elif ls_sub.x.shape[0]==0:
        print ('L8 x shape is 0')
        ls_sub = da.sel(y=slice(polary[1],polary[0]),x=slice(polarx[1],polarx[0]))
    print(ls_sub.shape)
    
    return ls_sub

##########################

def lsat_reproj(old_cs,new_cs,lbox):
    '''
    bbox comes out with the points out of order for making a polygon though pairs are correct. Order is 0,3,1,2 when done in normal projection. 
    Haven't tested for flipped grid.
    '''
    
    test_threshold = 0.5
    
    # Create a transform object to convert between coordinate systems
    inProj = Proj(init=old_cs)
    outProj = Proj(init=new_cs)
    
    ULX,LRY,LRX,ULY = lbox

    [lULX,lLRX], [lULY,lLRY] =  transform(inProj,outProj,[ULX,LRX], [ULY,LRY], always_xy=True)
    [cULX,cLRX], [cULY,cLRY] =  transform(outProj,inProj,[lULX,lLRX], [lULY,lLRY], always_xy=True)
    [lLLX,lURX], [lLLY,lURY] =  transform(inProj,outProj,[ULX,LRX], [LRY,ULY], always_xy=True)
    [cLLX,cURX], [cLLY,cURY] =  transform(outProj,inProj,[lLLX,lURX], [lLLY,lURY], always_xy=True)

    if LRY>ULY:
        bbox = [(lULX,lLLY),(lLLX,lULY),(lLRX,lURY),(lURX,lLRY)]
        print('lsat_reproj flipped orientation')
    else:
        bbox = [(lULX,lULY),(lLLX,lLLY),(lLRX,lLRY),(lURX,lURY)]
        print('lsat_reproj normal orientation')

    checkbox = np.array([cULX,cULY,cLRX,cLRY])
    if np.linalg.norm(checkbox - np.array([ULX,ULY,LRX,LRY])) > test_threshold:
        print(f"Round-trip transformation error 1 of {np.linalg.norm(checkbox - np.array([ULX,ULY,LRX,LRY]))}")
    checkbox = np.array([cLLX,cLLY,cURX,cURY])
    if np.linalg.norm(checkbox - np.array([ULX,LRY,LRX,ULY])) > test_threshold:
        print(f"Round-trip transformation error 2 of {np.linalg.norm(checkbox - np.array([ULX,LRY,LRX,ULY]))}")
    print (f'bbox={bbox}')
    print (f'lbox={lbox}')
    print (f'checkbox={checkbox}')
    
    return bbox,checkbox

##########################

def crop_xarray_dataarray_with_polygon(dataarray, polygon):
    """
    Crop an xarray.DataArray using a polygon.
    
    Parameters:
    - dataarray: xarray.DataArray with x and y coordinates.
    - polygon: Shapely Polygon object defining the crop area.
    
    Returns:
    - Cropped xarray.DataArray.
    """
    # Generate a 2D array of shapely Point objects for each grid point
    lon, lat = np.meshgrid(dataarray.x.values, dataarray.y.values)
    points = np.vectorize(Point)(lon, lat)
    
    # Create a mask where points within the polygon are True
    mask_func = np.vectorize(polygon.contains)
    mask = mask_func(points)
    
    # Convert the mask to an xarray.DataArray
    mask_da = xr.DataArray(mask, dims=["y", "x"], coords={"y": dataarray.y, "x": dataarray.x})
    
    # Apply the mask to the dataarray, cropping to the polygon
    # Use where method with drop=True to drop values outside the polygon
    cropped_dataarray = dataarray.where(mask_da, drop=True)
    
    return cropped_dataarray

##########################

def crosses_idl(coords):
    '''
    Determine if the set of coordinates crosses the International Dateline in a way that will mess up the creation of a polygon
    
    Variables:
    coords = list of lon, lat tuples
    
    Output:
    True or False
    '''
    
    for (lon1, lat1), (lon2, lat2) in zip(coords, coords[1:]):
        if abs(lon1 - lon2) >= 180:
            return True
    return False
```

```python
def _lon_to_ds_range(lon_da, target_lon):
    """Map target_lon into dataset's longitude range (0–360 or −180–180)."""
    lon_min = float(lon_da.min())
    lon_max = float(lon_da.max())
    if lon_max > 180 and lon_min >= 0:          # dataset uses 0..360
        targ = target_lon % 360
        lonr = lon_da % 360
    else:                                       # dataset uses −180..180
        targ = ((target_lon + 180) % 360) - 180
        lonr = ((lon_da + 180) % 360) - 180
    return lonr, targ

def plot_transect_at_lon(ds, target_lon=-101.7, tol_deg=0.05, ax=None):
    """
    Plot a 1D SSHA transect at a fixed longitude.

    Parameters
    ----------
    ds : xarray.Dataset with variables 'longitude','latitude','ssha'
    target_lon : float, degrees east (use negative for West, e.g., -101.7)
    tol_deg : keep only pixels within this |lon - target_lon| tolerance
    ax : optional matplotlib Axes to plot on
    """
    lon = ds['longitude']
    lat = ds['latitude']
    ssha = ds['ssha']

    # Put longitudes into a consistent range w.r.t. the target
    lonr, targ = _lon_to_ds_range(lon, target_lon)

    # Case A: 1D lon/lat on a rectilinear grid
    if lonr.ndim == 1 and lat.ndim == 1:
        # select nearest column at that longitude, then sort north→south
        col = ds.sel(longitude=targ, method='nearest')
        tran_lat = col['latitude']
        tran_ssha = col['ssha']
        # if latitude is a coord on the column, ensure descending (north→south)
        if tran_lat.ndim == 1:
            order = tran_lat.argsort()
            tran_lat = tran_lat.isel({tran_lat.dims[0]: order})
            tran_ssha = tran_ssha.isel({tran_ssha.dims[0]: order})
        # plot
        if ax is None:
            fig, ax = plt.subplots(figsize=(7,4))
        ax.plot(tran_lat, tran_ssha*100)
        ax.set_xlabel("Latitude (°N)")
        ax.set_ylabel("SSHA (cm)")
        ax.set_title(f"Transect at {abs(target_lon):.1f}°{'W' if target_lon<0 else 'E'}")
        return ax

    # Case B: 2D curvilinear lon/lat (typical for SWOT swath)
    # Find nearest-longitude pixel in each row
    row_dim, col_dim = ssha.dims  # e.g., ('num_lines','num_pixels')
    # distance to target lon in degrees
    dlon = np.abs(lonr - targ)
    j_star = dlon.argmin(dim=col_dim)                      # index of nearest col for each row
    ii = xr.DataArray(np.arange(ssha.sizes[row_dim]), dims=row_dim)

    tran_lon = lonr.isel({row_dim: ii, col_dim: j_star})
    tran_lat = lat.isel({row_dim: ii, col_dim: j_star})
    tran_ssha = ssha.isel({row_dim: ii, col_dim: j_star})

    # keep only points sufficiently close to the target longitude
    keep = np.isfinite(tran_ssha) & (np.abs(tran_lon - targ) <= tol_deg)
    tran_lat = tran_lat.where(keep, drop=True)
    tran_ssha = tran_ssha.where(keep, drop=True)

    # sort north→south (descending latitude)
    order = tran_lat.argsort()
    tran_lat = tran_lat.isel({tran_lat.dims[0]: order})
    tran_ssha = tran_ssha.isel({tran_ssha.dims[0]: order})

    # plot
    if ax is None:
        fig, ax = plt.subplots(figsize=(7,4))
    ax.plot(tran_lat, tran_ssha*100)
    ax.set_xlabel("Latitude (°N)")
    ax.set_ylabel("SSHA (cm)")
    ax.set_title(f"Transect at {abs(target_lon):.1f}°{'W' if target_lon<0 else 'E'}  (|Δlon| ≤ {tol_deg}°)")
    return ax
```

```python
def _normalize_lon(lon_da, target_lon):
    """Normalize longitudes so comparisons with target are consistent."""
    lon_min = float(np.nanmin(lon_da))
    lon_max = float(np.nanmax(lon_da))
    if (lon_max > 180) and (lon_min >= 0):         # dataset uses 0..360
        lonr = lon_da % 360
        targ = target_lon % 360
    else:                                          # dataset uses −180..180
        lonr = ((lon_da + 180) % 360) - 180
        targ = ((target_lon + 180) % 360) - 180
    return lonr, targ

def extract_transect_pixels(ds, target_lon=-102.3, mode="both", zero_tol_deg=1e-6, take_abs=False):
    """
    Collect pixel-center values (no interpolation) where a fixed meridian crosses the swath.

    Parameters
    ----------
    ds : xarray.Dataset with variables 'longitude','latitude','ssha'
    target_lon : float  (e.g., -102.3 for 102.3°W)
    mode : 'both' -> return both neighbors that straddle the meridian (denser)
           'nearest' -> return only the nearer neighbor per crossing (1 per row)
    zero_tol_deg : tolerance for treating a longitude as exactly on the meridian
    take_abs : if True, return |ssha| (magnitude). Default False = signed values.

    Returns
    -------
    pandas.DataFrame with columns:
      ['row','col','latitude','longitude','ssha_cm']
      sorted north→south (descending latitude)
    """
    lon = ds['longitude']; lat = ds['latitude']; ssha = ds['ssha']
    assert lon.ndim == lat.ndim == 2, "Expecting 2D curvilinear lon/lat for SWOT LR."
    row_dim, col_dim = ssha.dims

    lonr, targ = _normalize_lon(lon, target_lon)

    # differences from the meridian at adjacent columns
    d = lonr - targ
    left  = d.isel({col_dim: slice(None, -1)})
    right = d.isel({col_dim: slice(1,  None)})

    # valid where both sides (and both SSHA neighbors) are finite
    valid = (
        np.isfinite(left) & np.isfinite(right) &
        np.isfinite(ssha.isel({col_dim: slice(None, -1)})) &
        np.isfinite(ssha.isel({col_dim: slice(1, None)})) &
        np.isfinite(lat.isel({col_dim: slice(None, -1)})) &
        np.isfinite(lat.isel({col_dim: slice(1, None)})) &
        np.isfinite(lon.isel({col_dim: slice(None, -1)})) &
        np.isfinite(lon.isel({col_dim: slice(1, None)}))
    )

    # crossing if sign changes OR one side is essentially on the line
    cross = (
        ((left * right) < 0) |
        (np.abs(left) <= zero_tol_deg) |
        (np.abs(right) <= zero_tol_deg)
    ) & valid

    rows, jleft = np.where(cross.values)
    if rows.size == 0:
        return pd.DataFrame(columns=['row','col','latitude','longitude','ssha_cm'])

    # Build selections
    if mode == "both":
        # include both neighbors that straddle the meridian
        row_idx = np.concatenate([rows, rows])
        col_idx = np.concatenate([jleft, jleft + 1])
    elif mode == "nearest":
        # choose the neighbor whose center longitude is nearer the meridian
        abs_left  = np.abs(left.values)[rows, jleft]
        abs_right = np.abs(right.values)[rows, jleft]
        choose_right = (abs_right < abs_left)
        row_idx = rows
        col_idx = jleft + choose_right.astype(int)
    else:
        raise ValueError("mode must be 'both' or 'nearest'")

    lat_v  = lat.values[row_idx, col_idx]
    lon_v  = lon.values[row_idx, col_idx]          # keep in dataset’s original range for plotting
    ssha_v = ds['ssha'].values[row_idx, col_idx]   # in meters

    if take_abs:
        ssha_v = np.abs(ssha_v)

    df = pd.DataFrame({
        'row': row_idx,
        'col': col_idx,
        'latitude': lat_v,
        'longitude': lon_v,
        'ssha_cm': ssha_v * 100.0  # convert to cm
    }).dropna()

    # sort north → south
    df = df.sort_values('latitude').reset_index(drop=True)

    # attach time/granule tags if present
    if 'time' in ds.coords:
        try:
            df['time'] = pd.to_datetime(ds['time'].values)
        except Exception:
            pass
    if 'granule' in ds.attrs:
        df['granule'] = ds.attrs['granule']

    return df

def plot_transect_hits(ax, df, target_lon=-102.3, label='Meridian hits', color='k'):
    """Overlay the meridian and the discrete hit pixels on an existing map axis."""
    if df.empty:
        return
    # ax.plot(df['longitude'].values, df['latitude'].values, '.', ms=12, color=color, label=label)
    # draw the meridian itself as a thin line through map extent (visual aid)
    y0, y1 = ax.get_ylim()
    x_line = np.full(100, fill_value=float(target_lon))
    y_line = np.linspace(y0, y1, 100)
    ax.plot(x_line, y_line, '--', lw=1, color=color, alpha=0.7)
```

```python
def compute_relative_vorticity(ug, vg, lat):
    omega = 7.2921159e-05  # angular velocity of the Earth [rad/s]
    fc = 2*omega*np.sin(lat*np.pi/180.)
        # avoid zero near equator, bound fc by min val as 1.e-8
    f_coriolis = np.sign(fc)*np.maximum(np.abs(fc), 1.e-8)
    
    # ds = self._dset
        
    dx = 4000 # m
    dy = 4000 # m
        
    # du_dx, du_dy = np.gradient(ug, dx, edge_order=2)
    # dv_dx, dv_dy = np.gradient(vg, dx, edge_order=2)
    du_dy, du_dx = np.gradient(ug, dx, edge_order=2)
    dv_dy, dv_dx = np.gradient(vg, dx, edge_order=2)
        
    ksi = (dv_dx - du_dy)/f_coriolis
        
    return ksi
```

```python
def find_MODIS_SWOT(lonboundsC,latboundsC,ls_time):
        '''
        Finds the MODIS scene most closely coincident to a SWOT acquisition
        
        Variables: 
        ls_scene = xarray for one Landsat scene
        
        Outputs:
        mod_scene = xarray of MODIS SST image coincident in time with the Landsat scene
        granules[ind]['umm']['GranuleUR'] = modis file name
        min_time = the time difference between the Landsat image acquisition and chosen MODIS image
        '''
        
        # To convert from K to C
        KtoC = -273.15
    
        mbox = (lonboundsC[0],latboundsC[0],lonboundsC[1],latboundsC[1]) #east, south,west,north
        lsatpoly = Polygon([(lonboundsC[0],latboundsC[1]),(lonboundsC[0],latboundsC[0]),(lonboundsC[1],latboundsC[0]),(lonboundsC[1],latboundsC[1]),(lonboundsC[0],latboundsC[1])])
        
        # Get date/time for Landsat image and search for corresponding MODIS imagery  
        calc_dt = datetime.strptime(ls_time, '%Y-%m-%d %H:%M:%S').replace(tzinfo=pytz.UTC)
        # start_dt = (calc_dt + timedelta(days=-0.5)).strftime('%Y-%m-%d %H:%M:%S')
        # end_dt = (calc_dt + timedelta(days=0.5)).strftime('%Y-%m-%d %H:%M:%S')
        # Set to beginning and end of the same day
        start_dt = calc_dt.replace(hour=0, minute=0, second=0).strftime('%Y-%m-%d %H:%M:%S')
        end_dt = calc_dt.replace(hour=23, minute=59, second=59).strftime('%Y-%m-%d %H:%M:%S')
        
        # Gather all files from search location from Terra and Aqua for the same day as the Landsat image
        granules = earthaccess.search_data(
            short_name='MODIS_T-JPL-L2P-v2019.0',
            bounding_box=mbox,
            # Day of a landsat scene to day after - searches day of only
            temporal=(start_dt,end_dt)
        )
        granules2 = earthaccess.search_data(
            short_name='MODIS_A-JPL-L2P-v2019.0', #MODIS_AQUA_L3_SST_THERMAL_DAILY_4KM_NIGHTTIME_V2019.0
            bounding_box=mbox,
            # Day of a landsat scene to day after - searches day of only
            temporal=(start_dt,end_dt)
        )
        granules = granules + granules2
        print (f'{len(granules)} TOTAL MODIS granules')
        
        # Filter granules by spatial overlap
        best_grans = []
        granule_times = []
        for granule in granules:
            try:
                gpolys = granule['umm']['SpatialExtent']['HorizontalSpatialDomain']['Geometry']['GPolygons']
                for gpoly in gpolys:
                    map_points = [(p['Longitude'], p['Latitude']) for p in gpoly['Boundary']['Points']]
                    pgon = Polygon(map_points)
                    if lsatpoly.difference(pgon).area / lsatpoly.area <= 20.0:
                        if not crosses_idl(map_points):
                            best_grans.append(granule)
                            granule_times.append(pd.to_datetime(granule['umm']['TemporalExtent']['RangeDateTime']['BeginningDateTime']))
                            break
            except Exception as e:
                print(f"Skipping granule due to error: {e}")
        
        print(f'{len(best_grans)} remaining MODIS granules')
        
        paths = earthaccess.open(best_grans)
        
        # Open all remaining granules as xarray datasets
        mod_scenes = []
        filenames = []
        for granule, gtime in zip(paths, granule_times):
            try:
                ds = xr.open_dataset(granule,decode_timedelta=True)
                ds = ds.rio.write_crs("epsg:4326", inplace=True)
            
                # Add time_dif as a variable (in seconds)
                time_diff_seconds = (gtime - calc_dt).total_seconds()
                ds["time_dif"] = xr.DataArray(time_diff_seconds, attrs={"units": "seconds", "description": "Time difference between MODIS and Landsat"}, dims=())
                ds['sea_surface_temperature'] = (ds.sea_surface_temperature + KtoC).where(ds.sea_surface_temperature + KtoC > -55.5, np.nan)

                # Add time dimension
                ds = ds.expand_dims({"time": [gtime]})  # creates a new 'time' dimension
                ds = ds.assign_coords(time=("time", [gtime]))
            
                mod_scenes.append(ds)
                filenames.append(granule.full_name[-51:])
            except Exception as e:
                print(f"Failed to open MODIS file: {e}")
        
        if not mod_scenes:
            print("⚠️ No MODIS datasets successfully opened.")
        
        mod_combined = xr.concat(mod_scenes, dim="time", combine_attrs="override")
    
        return mod_combined, paths
    
        # # Find MODIS image closest in time to each Landsat image
        # # Make Landsat datetime timezone aware (UTC)
        # Mdates = [pd.to_datetime(granule['umm']['TemporalExtent']['RangeDateTime']['BeginningDateTime']) for granule in best_grans]
        # # ind = Mdates.index(min( Mdates, key=lambda x: abs(x - pytz.utc.localize(ls_time))))
        # ind = Mdates.index(min( Mdates, key=lambda x: abs(x - calc_dt)))
        # time_dif = abs(Mdates[ind] - calc_dt)
        # print(f'Time difference between MODIS and Landsat: {time_dif}')
    
        # mod_scene = xr.open_dataset(earthaccess.open(best_grans[ind:ind+1])[0])
        # mod_scene = mod_scene.rio.write_crs("epsg:4326", inplace=True) 
        
        # return mod_scene, granules[ind]['umm']['GranuleUR'],time_dif
```

```python
start_dt = '2020-01-01'
end_dt   = '2024-12-31'

clim_ulat = 42
clim_llat = 32
clim_llon = -80 
clim_rlon = -40

mbox = (clim_llon, clim_llat, clim_rlon, clim_ulat)

# Search both sensors (monthly products) over the bounding box/time 
granules_aqua = earthaccess.search_data(
    short_name='MODIS_AQUA_L3_SST_THERMAL_MONTHLY_4KM_NIGHTTIME_V2019.0',
    bounding_box=mbox,
    temporal=(start_dt, end_dt),
)
granules_terra = earthaccess.search_data(
    short_name='MODIS_TERRA_L3_SST_THERMAL_MONTHLY_4KM_NIGHTTIME_V2019.0',
    bounding_box=mbox,
    temporal=(start_dt, end_dt),
)

granules = granules_aqua + granules_terra
print(f'{len(granules)} TOTAL MODIS monthly granules found')
paths = earthaccess.open(granules) 

# Extract each granule's nominal month
g_month = []
for g in granules:
    gtime = pd.to_datetime(g['umm']['TemporalExtent']['RangeDateTime']['BeginningDateTime'])
    g_month.append(gtime.to_period('M')) 

# Helper to open and standardize a single granule to Celsius and grid metadata
def open_sst(paths_idx):
    ds = xr.open_dataset(paths[paths_idx], decode_timedelta=True)
    # Ensure CRS (lat/lon WGS84). No reprojection is performed here.
    ds = ds.rio.write_crs("epsg:4326", inplace=True)
    # Drop junk values 
    sst_c = ds['sst'].where(ds['sst'] > -55.5, np.nan)

    return sst_c



# Iterate month, average Terra + Aqua within each month
months = pd.period_range(start=start_dt, end=end_dt, freq='M')  # inclusive monthly periods
monthly_means = []

for mo in tqdm(months, desc='Building monthly means'):
    # Extract path indices of granules that belong to this calendar month
    idxs = [i for i, pm in enumerate(g_month) if pm == mo]
    if len(idxs) == 0:
        print(f'No granules for {mo}; skipping')
        continue

    # Open all granules for this month (usually 1 Aqua + 1 Terra)
    month_fields = []
    for i in idxs:
        try:
            xx = open_sst(i)
            # Checking to make sure has the same lat/lon grid as the rest of the images, otherwise will cause concatenation issues
            # if np.around(xx.lat[-1].values.item(),5)==-89.97918:
            month_fields.append(xx)
            # else: 
            #     print(f'{mo} {i}; skipping')
            #     continue
        except Exception as e:
            print(f'Failed opening granule {i} for {mo}: {e}')

    if len(month_fields) == 0:
        print(f'No usable granules for {mo}; skipping')
        continue

    # Stack along a synthetic dimension and mean them (Terra+Aqua monthly average)
    sst_month = xr.concat(month_fields, dim='source').mean('source', skipna=True)

    # Crop to bounding box
    sst_month = sst_month.sel(
        lat=slice(clim_ulat,clim_llat),
        lon=slice(clim_llon, clim_rlon))

    if 0 in sst_month.shape:
        print(f"{mo}, skipping")
        continue

    # Add a time coordinate at the month start (MS)
    t0 = mo.to_timestamp(how='start')  # pandas.Timestamp at 1st of month
    sst_month = sst_month.expand_dims(time=[t0])

    monthly_means.append(sst_month)

# Combine all months into one DataArray (time, lat, lon)
if len(monthly_means) == 0:
    raise RuntimeError("No monthly data was assembled. Check search bounds or files.")
sst_monthly = xr.concat(monthly_means, dim='time')
sst_monthly.attrs.update({
    'long_name': 'Monthly mean SST (Terra+Aqua, nighttime, 4km)',
    'units': 'degC'
})

# Climatologies
# 12-month climatologies
sst_clim_12 = (
    sst_monthly
    .groupby('time.month')
    .mean('time', skipna=True)
    .rename(month='month_of_year')
)
sst_clim_12.attrs.update({
    'long_name': 'Calendar-month climatology of SST (degC)'
})

# --- Unweighted seasonal climatology (DJF, MAM, JJA, SON) ---
# sst_monthly: DataArray (time, lat/lon or y/x) of monthly means
sst_clim_seasonal = (
    sst_monthly
    .groupby('time.season')             # seasons: DJF, MAM, JJA, SON
    .mean('time', skipna=True)
).sel(season=['DJF','MAM','JJA','SON'])

sst_clim_seasonal.name = 'sst_climatology_seasonal'
sst_clim_seasonal.attrs.update({
    'long_name': 'Seasonal climatology of SST (degC)',
    'note': 'Unweighted by month length'
})

# --- Days-weighted seasonal climatology (better for monthly means) ---
# Weight each month by its number of days, then compute seasonal means
w = xr.DataArray(sst_monthly['time'].dt.days_in_month, dims=['time'])

# Sum (sst * days) over time within each season, then divide by total days in that season
num = (sst_monthly * w).groupby('time.season').sum('time', skipna=True)
den = w.groupby(sst_monthly['time'].dt.season).sum()

sst_clim_seasonal_weighted = (num / den).sel(season=['DJF','MAM','JJA','SON'])
sst_clim_seasonal_weighted.name = 'sst_climatology_seasonal_weighted'
sst_clim_seasonal_weighted.attrs.update({
    'long_name': 'Seasonal climatology of SST (degC)',
    'note': 'Weighted by days in month'
})


# All-time climatology over the full window (single 2D field)
sst_clim_all = sst_monthly.mean('time', skipna=True)
sst_clim_all.name = 'sst_climatology'
sst_clim_all.attrs.update({
    'long_name': f'All-time mean SST over {start_dt}–{end_dt} (degC)'
})

# Package into a Dataset
clim_ds = xr.Dataset(
    data_vars=dict(
        sst_monthly = sst_monthly,              # (time, y, x) or (time, lat, lon)
        sst_climatology_seasonal = sst_clim_seasonal,
        sst_climatology_seasonal_wted = sst_clim_seasonal_weighted,
        sst_climatology_12mo = sst_clim_12,     # (month_of_year, y, x)
        sst_climatology_alltime = sst_clim_all  # (y, x)
    )
)

# Save to netcdf:
sst_monthly.to_netcdf('modis_sst_monthly_terra_aqua_4km_night_2019_2024.nc')
sst_clim_12.to_netcdf('modis_sst_climatology_12mo_4km_night_2019_2024.nc')
sst_clim_seasonal_weighted.to_netcdf('modis_sst_climatology_seas_4km_night_2019_2024.nc')
sst_clim_all.to_netcdf('modis_sst_climatology_alltime_4km_night_2019_2024.nc')

print(clim_ds)
```
